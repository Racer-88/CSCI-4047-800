```python
# Code here
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Income Data.csv")
print(df.head())
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
# Code here

mean = df[['income', 'education', 'prestige']].mean()
minv = df[['income', 'education', 'prestige']].min()
maxv = df[['income', 'education', 'prestige']].max()
median = df[['income', 'education', 'prestige']].median()
qone = df[['income', 'education', 'prestige']].quantile(0.25)
qthree = df[['income', 'education', 'prestige']].quantile(0.75)

print("minv:\n", minv)
print("qone:\n", qone)
print("median:\n", median)
print("Mean:\n", mean)
print("qthree:\n", qthree)
print("maxv:\n", maxv)
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
# Code here
plt.hist(df['prestige'], bins=12, color='orange', alpha=0.5)
plt.show()
```

```python
plt.hist(df['education'], bins=12, color='blue', alpha=0.5)
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
# Code here
plt.scatter(x=df['income'], y = df['education'], alpha=0.3)
plt.title("Income compared to Education")
plt.xlabel('income')
plt.ylabel("education")
plt.show()
```

```python
plt.scatter(x=df['education'], y = df['prestige'], alpha=0.3)
plt.title("Education compared to Prestige")
plt.xlabel('education')
plt.ylabel("prestige")
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
# Code here
standard_dev = df['income'].std()
print(f"Standard Deviation:  {standard_dev:.2f}")
variance = standard_dev ** 2
print(f"Variance:            {variance:.2f}")
```
